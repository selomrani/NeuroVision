```python
import cv2
import numpy as np
import os

from numpy.distutils.fcompiler import none
```

we start by filtering out the corrupted files  and resizing the correct ones to a fixed format of 224 by 224

```python
images = []
labels = []
skipped = []
classes = [{
    'class_name' : "glioma"
},
    {
    'class_name' : "meningioma"
},
    {
    'class_name' : "notumor"
},
    {
    'class_name' : "pituitary"
}
]

for classo in classes :
    folder = f"../Data/raw/{classo['class_name']}"
    for file_name in os.listdir(folder) :
        image_path = os.path.join(folder,file_name)
        try:
            img = cv2.imread(image_path)
            if img is None :
                raise ValueError("Image corrupted")
            img = cv2.resize(img, (224, 224))
            images.append(img)
            labels.append(classo['class_name'])
        except Exception as exc:
             skipped.append((image_path, str(exc)))
             continue
print(images)
```

Now we map each image with its respective class / lable as a numpy array

```python
X = np.array(labels)
Y = np.where(images)
```
